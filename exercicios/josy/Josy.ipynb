Exercício 1 - Variáveis e tipos

```python
# Solicitando os dados ao usuário
nome = input("Digite seu nome: ")
ano_nascimento = int(input("Digite seu ano de nascimento: "))
cidade = input("Digite a cidade onde você mora: ")

# Calculando a idade aproximada (usando 2026 como ano atual)
ano_atual = 2026
idade = ano_atual - ano_nascimento

# Exibindo a frase com f-string
print(f"{nome}, {idade} anos, mora em {cidade}.")
```

Exercício 2 - Condicionais

```python
numero = int(input("Digite um número: "))

# Verificando se é positivo, negativo ou zero
if numero > 0:
    sinal = "positivo"
elif numero < 0:
    sinal = "negativo"
else:
    sinal = "zero"

# Verificando se é par ou ímpar usando o resto da divisão (%)
if numero % 2 == 0:
    paridade = "par"
else:
    paridade = "ímpar"

print(f"O número {numero} é {sinal} e é {paridade}.")
```

Exercício 3 - Loops

```python
numero = int(input("Digite um número para ver a tabuada: "))

print(f"\n--- Tabuada do {numero} usando FOR ---")
# O range vai de 1 a 10 (o limite superior 11 não é incluído)
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")

print(f"\n--- Tabuada do {numero} usando WHILE ---")
contador = 1
while contador <= 10:
    print(f"{numero} x {contador} = {numero * contador}")
    contador += 1
```

Exercício 4 - Listas e dicionários

```python
# Criando o dicionário do produto
produto = {
    "nome": "Notebook",
    "preco": 3500.50,
    "quantidade_em_estoque": 12
}

# Calculando o valor total (preço * quantidade)
valor_total = produto["preco"] * produto["quantidade_em_estoque"]

# Exibindo o resultado formatado
print(f"Produto: {produto['nome']}")
print(f"Valor total em estoque: R$ {valor_total:.2f}")
```

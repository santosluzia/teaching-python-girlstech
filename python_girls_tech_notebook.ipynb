### 6. Funções — exercício integrador da Semana 1

```python
def analisar_notas(notas):
    pares = [n for n in notas if n % 2 == 0]
    impares = [n for n in notas if n % 2 != 0]
    media = sum(notas) / len(notas)
    return {"pares": pares, "impares": impares, "media": media}

resultado = analisar_notas([8, 7, 9, 6, 10])
print(resultado)
```

### 8. Tratamento de exceções

```python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Não é possível dividir por zero"

print(dividir(10, 2))
print(dividir(10, 0))
```

### 10. Programação Orientada a Objetos

```python
class ContaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, valor):
        self.saldo += valor

    def sacar(self, valor):
        if valor > self.saldo:
            print("Saldo insuficiente")
        else:
            self.saldo -= valor

conta = ContaBancaria("Ana")
conta.depositar(100)
conta.sacar(30)
print(f"Saldo de {conta.titular}: R$ {conta.saldo}")
```

### 12. Automação — organizando arquivos por extensão

```python
import os
import shutil

os.makedirs("meus_arquivos", exist_ok=True)
for nome_arquivo in ["relatorio.txt", "planilha.csv", "notas.txt"]:
    caminho = os.path.join("meus_arquivos", nome_arquivo)
    with open(caminho, "w") as f:
        f.write("conteúdo de exemplo")

for arquivo in os.listdir("meus_arquivos"):
    if os.path.isfile(os.path.join("meus_arquivos", arquivo)):
        ext = arquivo.split(".")[-1]
        pasta_destino = os.path.join("meus_arquivos", ext)
        os.makedirs(pasta_destino, exist_ok=True)
        shutil.move(os.path.join("meus_arquivos", arquivo), pasta_destino)

for raiz, pastas, arquivos in os.walk("meus_arquivos"):
    for arquivo in arquivos:
        print(os.path.join(raiz, arquivo))
```

### 13. Projeto prático final — gerenciador de gastos pessoais

```python
import csv

with open("gastos.csv", "w", newline="", encoding="utf-8") as f:
    escritor = csv.writer(f)
    escritor.writerow(["data", "categoria", "valor"])
    escritor.writerows([
        ["2026-09-01", "Alimentação", 50],
        ["2026-09-03", "Transporte", 20],
        ["2026-09-05", "Alimentação", 30],
        ["2026-09-10", "Lazer", 40],
    ])

totais = {}
with open("gastos.csv", "r", encoding="utf-8") as f:
    leitor = csv.DictReader(f)
    for linha in leitor:
        categoria = linha["categoria"]
        valor = float(linha["valor"])
        totais[categoria] = totais.get(categoria, 0) + valor

with open("resumo_gastos.txt", "w", encoding="utf-8") as f:
    for categoria, total in totais.items():
        linha_resumo = f"{categoria}: R$ {total:.2f}"
        print(linha_resumo)
        f.write(linha_resumo + "\n")
```
